```python
import pandas as pd
import numpy as np
from collections import Counter
import random

def analisar_lotofacil(arquivo_excel):
    # Ler a planilha
    try:
        df = pd.read_excel(arquivo_excel, sheet_name='LOTOFÁCIL')
        print("Planilha carregada com sucesso!")
    except Exception as e:
        print(f"Erro ao carregar a planilha: {e}")
        return

    # Coletar todos os números sorteados
    numeros_sorteados = []
    colunas_bolas = [f'Bola{i}' for i in range(1, 16)]

    for _, row in df.iterrows():
        for coluna in colunas_bolas:
            if pd.notna(row[coluna]):
                numeros_sorteados.append(int(row[coluna]))

    # Análise de frequência
    contador = Counter(numeros_sorteados)

    # Números mais sorteados
    print("\n" + "="*60)
    print("NÚMEROS MAIS SORTEADOS")
    print("="*60)
    mais_sorteados = contador.most_common(15)
    for numero, frequencia in mais_sorteados:
        print(f"Número {numero:2d}: {frequencia:3d} vezes ({frequencia/len(df)*100:.2f}%)")

    # Números menos sorteados
    print("\n" + "="*60)
    print("NÚMEROS MENOS SORTEADOS")
    print("="*60)
    menos_sorteados = contador.most_common()[-15:]
    for numero, frequencia in menos_sorteados:
        print(f"Número {numero:2d}: {frequencia:3d} vezes ({frequencia/len(df)*100:.2f}%)")

    # Preparar sugestões baseadas em estatísticas
    print("\n" + "="*60)
    print("SUGESTÕES DE JOGOS - MAIOR CHANCE DE ACERTOS")
    print("="*60)

    # Jogos com maior chance (baseados nos números mais sorteados)
    numeros_mais_frequentes = [num for num, _ in mais_sorteados]

    for i in range(20):
        jogo = sorted(random.sample(numeros_mais_frequentes, 15))
        print(f"Jogo {i+1:2d}: {jogo}")

    print("\n" + "="*60)
    print("SUGESTÕES DE JOGOS - MENOR CHANCE DE ACERTOS")
    print("="*60)

    # Jogos com menor chance (baseados nos números menos sorteados)
    numeros_menos_frequentes = [num for num, _ in menos_sorteados]

    for i in range(20):
        jogo = sorted(random.sample(numeros_menos_frequentes, 15))
        print(f"Jogo {i+1:2d}: {jogo}")

    # Estatísticas adicionais
    print("\n" + "="*60)
    print("ESTATÍSTICAS GERAIS")
    print("="*60)
    print(f"Total de concursos analisados: {len(df)}")
    print(f"Período analisado: {df['Data Sorteio'].min()} até {df['Data Sorteio'].max()}")
    print(f"Total de números sorteados: {len(numeros_sorteados)}")

    # Distribuição por faixa de números
    print("\nDistribuição por faixas:")
    faixas = {
        "1-5": sum(1 for num in numeros_sorteados if 1 <= num <= 5),
        "6-10": sum(1 for num in numeros_sorteados if 6 <= num <= 10),
        "11-15": sum(1 for num in numeros_sorteados if 11 <= num <= 15),
        "16-20": sum(1 for num in numeros_sorteados if 16 <= num <= 20),
        "21-25": sum(1 for num in numeros_sorteados if 21 <= num <= 25)
    }

    for faixa, quantidade in faixas.items():
        percentual = (quantidade / len(numeros_sorteados)) * 100
        print(f"Faixa {faixa}: {quantidade:4d} vezes ({percentual:.2f}%)")

def gerar_jogos_estrategicos(arquivo_excel, num_jogos=20):
    """Gera jogos estratégicos baseados em diferentes abordagens"""

    df = pd.read_excel(arquivo_excel, sheet_name='LOTOFÁCIL')

    # Coletar todos os números sorteados
    numeros_sorteados = []
    colunas_bolas = [f'Bola{i}' for i in range(1, 16)]

    for _, row in df.iterrows():
        for coluna in colunas_bolas:
            if pd.notna(row[coluna]):
                numeros_sorteados.append(int(row[coluna]))

    contador = Counter(numeros_sorteados)

    # Estratégia 1: Números mais frequentes
    mais_frequentes = [num for num, _ in contador.most_common(25)]

    # Estratégia 2: Distribuição equilibrada por faixas
    faixas = {
        "1-5": [1, 2, 3, 4, 5],
        "6-10": [6, 7, 8, 9, 10],
        "11-15": [11, 12, 13, 14, 15],
        "16-20": [16, 17, 18, 19, 20],
        "21-25": [21, 22, 23, 24, 25]
    }

    print("\n" + "="*70)
    print("JOGOS ESTRATÉGICOS - DISTRIBUIÇÃO EQUILIBRADA")
    print("="*70)

    for i in range(num_jogos):
        jogo = []
        # Pegar 3 números de cada faixa
        for faixa_numeros in faixas.values():
            jogo.extend(random.sample(faixa_numeros, 3))

        # Ordenar o jogo
        jogo.sort()
        print(f"Jogo {i+1:2d}: {jogo}")

    # Estratégia 3: Números que não saem há mais tempo (simulação)
    print("\n" + "="*70)
    print("JOGOS ESTRATÉGICOS - MISTURA DE FREQUÊNCIAS")
    print("="*70)

    for i in range(num_jogos):
        # Misturar números de diferentes frequências
        jogo = []
        jogo.extend(random.sample(mais_frequentes[:8], 8))  # 8 números mais frequentes
        jogo.extend(random.sample(mais_frequentes[8:17], 4))  # 4 números de frequência média
        jogo.extend(random.sample(mais_frequentes[17:], 3))  # 3 números menos frequentes

        jogo.sort()
        print(f"Jogo {i+1:2d}: {jogo}")

# Executar a análise
if __name__ == "__main__":
    arquivo = "Lotofácil.xlsx"

    print("ANÁLISE DA LOTOFÁCIL")
    print("="*60)

    # Análise principal
    analisar_lotofacil(arquivo)

    # Jogos estratégicos adicionais
    gerar_jogos_estrategicos(arquivo)

    print("\n" + "="*60)
    print("DICAS PARA APOSTAS:")
    print("="*60)
    print("1. Os jogos com maior chance usam números que saem com mais frequência")
    print("2. Os jogos com menor chance usam números que saem com menos frequência")
    print("3. Para maior diversificação, misture números de diferentes faixas")
    print("4. Lembre-se: a loteria é um jogo de sorte!")
```
