```python
#Você deverá criar um programa que:
#Pergunte ao aluno:
#Quantas horas ele estudou por dia em cada dia da semana.
#Quantos exercícios ele realizou no total.
#Quantas aulas ele assistiu (de um total de 5 disponíveis).
#Se ele participou de discussões ou fóruns (responder “sim” ou “não”).
#Com base nessas informações, o programa deve calcular uma pontuação final de 0 a 1, considerando:
#Esforço diário (50%) — média de horas estudadas / 10
#Participação (30%) — proporção de aulas assistidas e participação em fóruns
#Prática (20%) — número de exercícios realizados dividido por um valor máximo (ex: 50)
#O programa deve:
#Usar loops for para coletar dados de cada dia da semana.
#Usar while para garantir que as respostas do usuário sejam válidas (ex: “sim” ou “não”).
#Usar condicionais e operadores lógicos (and, or, not) para verificar níveis de engajamento.
#O programa deve imprimir:
#A nota final (entre 0 e 1).
#Uma mensagem de feedback com base no resultado:
#>= 0.8 → “Excelente! Continue assim!”
#0.5–0.79 → “Bom, mas há espaço para melhorar.”
#< 0.5 → “Atenção! É preciso mais dedicação.”
#Dica para o aluno
#Você precisará pesquisar:
#Como garantir entradas válidas com while
#Como usar operadores lógicos (and, or, not)
#Como normalizar valores (para trazer tudo à faixa de 0–1)
#Como formatar saídas (print(f"Nota: {nota:.2f}"))
testar = bool(True)
while testar == True:
  horas_estudadas = float(input("Quantas horas você estudou por dia em cada dia da semana? "))
  if horas_estudadas < 0:
    print("Por favor, insira um valor válido.")
    continue
  exercicios_realizados = int(input("Quantos exercícios você realizou (de um total de 50 disponíveis)? "))
  if exercicios_realizados < 0:
    print("Por favor, insira um valor válido.")
    continue
  aulas_assistidas = int(input("Quantas aulas você assistiu (de um total de 5 disponíveis)? "))
  if aulas_assistidas < 0 or aulas_assistidas > 5:
    print("Por favor, insira um valor válido.")
    continue
  participacao_foruns = input("Você participou de discussões ou fóruns? (responda com sim ou não) ")
  if participacao_foruns != "sim" and participacao_foruns != "não":
    print("Por favor, insira uma resposta válida.")
    continue
  testar = False
  if horas_estudadas < 0 or exercicios_realizados < 0 or aulas_assistidas < 0:
    print("Atenção! É preciso mais dedicação.")
else:
  nota_final = (horas_estudadas / 10) * 0.5 + (exercicios_realizados / 50) * 0.2 + (aulas_assistidas / 5) * 0.3
  if participacao_foruns == "sim":
    nota_final += 0.1
  elif participacao_foruns == "não":
    nota_final -= 0.1
  if nota_final > 1.0:
    nota_final = 1.0
  if nota_final < 0.0:
    nota_final = 0.0
  if nota_final < 0.5:
    print(f"Nota: {nota_final:.2f} - Atenção! É preciso mais dedicação.")
  if nota_final >= 0.8:
    print(f"Nota: {nota_final:.2f} - Excelente! Continue assim!")
  elif nota_final >= 0.5 and nota_final <= 0.79:
    print(f"Nota: {nota_final:.2f} - Bom, mas há espaço para melhorar.")

```
