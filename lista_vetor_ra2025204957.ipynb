```python
#Tarefa em sala de aula individual
#Questão 1: em uma lista (array), verificar se cada elemento é par ou ímpar conforme abaixo:
#vet = [12,-1,8,0,4,23,12,15,-1,13]
#Imprimir resultado: o vetor/lista { num } de indice {x} é {par/impar}
#Dica:
#Podemos usar
#enumerate(vet) → retorna índice e valor em cada iteração.
#Exemplo: (0, 12), (1, -1), (2, 8) etc.
#Ou  range(len(vet)) → gera uma sequência de índices de 0 até len(vet) - 1.
#A cada passo, acessamos o número com vet[i]
#
#Questão 2:  Ler um conjunto de números reais armazenados em uma lista (array) e calcular o quadrado dos componentes deste array armazenando este valor em outra lista (array)
#numero = [3,-4,8,5,7,1,-2]
#A ideia é:
#Ler (ou definir) a lista original de números.
#Calcular o quadrado de cada elemento.
#Armazenar os resultados em outra lista.
#Mostrar as duas listas.
#Dica:
#Podemos fazer por potência ( x ** 2) ou multiplicando por ele mesmo (x ** x)
#Podemos usar também list comprehension como apresentado em aula
#Usando map() -> map(lambda )
#
#Questão 3:  Crie um algoritmo que leia um vetor de salários de 8 posições (fazer input float dos salários criando os arrays/listas), calcular e imprimir o vetor com os novos saláros reajustados em 17% (aumento de 17%)
#Vamos fazer passo a passo:
#Ler 8 salários com input() (armazenar em uma lista).
#Calcular o reajuste de 17%.
#Armazenar os novos salários em outra lista.
#Mostrar os dois vetores — original e reajustado.
#
#Questão 4: Faça um algoritmo que carregue um array com 15 elementos inteiros e verifique a existência de elementos iguais a 30 mostrando as posições em que esses elementos aparecem.
#array = [7,24, -1,30,2, ... , 15,30,0]
#Dica: Usar laço com if para verificar as ocorrências
#Como funciona o map() juntamente com o lambda?
#map() é uma função que aplica outra função a todos os elementos de uma lista (ou outro iterável).
#map(funcao, iteravel)
#funcao: o que você quer aplicar (ex: dobrar, somar, converter, etc.)
#iteravel: a lista, tupla ou conjunto de dados onde a função será aplicada
#-> Ele retorna um objeto map, que normalmente transformamos em lista com list()
#
#O que é lambda?
#lambda é uma função anônima, ou seja, uma função sem nome, usada para coisas simples e rápidas.
#lambda x: x * 2
#é o mesmo que:
#def dobrar(x):
#    return x * 2
#Juntando map() + lambda
#
#numeros = [2, 4, 6, 8]
#dobrados = list(map(lambda x: x * 2, numeros))
#print(dobrados)

#RESOLUÇÃO
# QUESTÃO 1

vet = [12,-1,8,0,4,23,12,15,-1,13]
par = []
impar = []
for x, num in enumerate(vet):
  if  num % 2 == 0:
    print(f"O vetor/lista {vet} de indice {x} é par")
    par.append(num)
  else:
    print(f"O vetor/lista {vet} de indice {x} é ímpar")
    impar.append(num)
print(f"Os números pares são: {par}")
print(f"Os números ímpares são: {impar}")


```

```python
#Questão 3:  Crie um algoritmo que leia um vetor de salários de 8 posições (fazer input float dos salários criando os arrays/listas), calcular e imprimir o vetor com os novos saláros reajustados em 17% (aumento de 17%)
#Vamos fazer passo a passo:
#Ler 8 salários com input() (armazenar em uma lista).
#Calcular o reajuste de 17%.
#Armazenar os novos salários em outra lista.
#Mostrar os dois vetores — original e reajustado.
#RESOLUÇÃO
# QUESTÃO 3
salarios = []
novo_salario = []
for i in range(8):
    while True:
        try:
          salario = float(input("Digite o salário: "))
          if salario > 0:
            salarios.append(salario)
            salario_corrigido = salario * 1.17
            novo_salario.append(salario_corrigido)
            break
          else:
            print("Digite um valor válido")
        except ValueError:
          print("Digite um valor válido")
print(f"Os salários originais são: {salarios}")
print(f"Os salários reajustados são: {novo_salario}")
```
