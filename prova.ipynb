```python
# Gerar matriz ano-mês-dia

anos = range(2010, 2026)   # 2010 até 2025
meses = range(1, 13)       # 1 até 12
dias = range(1, 31 + 1)    # 1 até 31

matriz = []

for ano in anos:
    for mes in meses:
        for dia in dias:
            matriz.append({
                "ano": ano,
                "mes": mes,
                "dia": dia
            })

# Exemplo: imprimir os 10 primeiros itens
for item in matriz[:10]:
    print(item)

print(f"\nTotal de registros: {len(matriz)}")
```
