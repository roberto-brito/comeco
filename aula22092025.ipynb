```python
# Inicializando o acumulador
soma = 0
contador = 1

# Estrutura de repetição
while contador <= 5:
  numero = int(input(f"Digite o {contador}º número: "))
  soma += numero
  contador +=1
# Exibe o resultado final
print(f"A soma dos números é: {soma} ")
```

```python
# Número secreto
numero_secreto=7
tentativas = 0

# Estrutura de repetição com teste no início
while True:
  palpite = int(input("Tente adivinhar o número secreto: "))
  tentativas +=1
  if palpite == numero_secreto:
    print(f"Parabéns! Você acertou o número secreto em {tentativas} tentativas.")
    break
  else:
    print("Palpite incorreto. Tente novamente.")
```

```python
# Número secreto
numero_secreto=7
tentativas = 0

# Estrutura de repetição com teste no início
while True:
  palpite = int(input("Tente adivinhar o número secreto: "))
  tentativas +=1
  if tentativas == 3:
    print("Você atingiu o limite de tentativas. O jogo será encerrado.")
    break
  elif palpite == numero_secreto:
    print(f"Parabéns! Você acertou o número secreto em {tentativas} tentativas.")
    break
  else:
    print("Palpite incorreto. Tente novamente.")
```
